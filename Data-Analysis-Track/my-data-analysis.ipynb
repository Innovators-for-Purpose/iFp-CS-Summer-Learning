```python
print(df.isna().sum()) #shows missing responses
```

```python
average_stress = df["Perceived_Stress_Score"].mean()
print(average_stress)
```

```python
average_GPA = df["Academic_Performance_GPA"].mean() 
print(average_GPA)
# calculates the most probable answer based on what's already there
```

```python
df["Perceived_Stress_Score"] = df["Perceived_Stress_Score"].fillna(average_stress)
df["Academic_Performance_GPA"] = df["Academic_Performance_GPA"].fillna(average_GPA)
#fills missing responses
```

```python
age_column = df["Age"]#1
usage_column = df["Daily_Usage_Hours"]#2
weekend_hours_column = df["Weekend_Extra_Hours"]#3
sleep_hours_column = df["Sleep_Duration_Hours"]#4
sleep_quality_column = df["Sleep_Quality_Score"]#5
stress_score_column = df["Perceived_Stress_Score"]#6
mental_health_column = df["Mental_Health_Index"]#7
GPA_column = df["Academic_Performance_GPA"]#8
```

```python
#[.describe]= statistics of the column named 
#(only used for numerical data)
print(age_column.describe())
print(usage_column.describe())
print(weekend_hours_column.describe())
print(sleep_hours_column.describe())
print(sleep_quality_column.describe())
print(stress_score_column.describe())
print(mental_health_column.describe())
print(GPA_column.describe())
```
