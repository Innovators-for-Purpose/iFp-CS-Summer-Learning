```python
two_conditions = df[(df["Daily_Usage_Hours"] > 8) & (df["Sleep_Duration_Hours"] < 6)]
print(two_conditions)
display(two_conditions) # This will display the rows of the DataFrame where the daily usage hours are greater than 8 and the sleep duration hours are less than 6
```

```python
print(two_conditions.shape[0]) # This will print the number of rows that meet both conditions
```

```python
three_conditions = df[(df["Daily_Usage_Hours"] > 6) & (df["Sleep_Duration_Hours"] < 7)]
print(three_conditions)
display(three_conditions) # This will display the rows of the DataFrame where the daily usage hours are greater than 6 and the sleep duration hours are less than 7
```

```python
four_conditions = df[(df["Daily_Usage_Hours"] > 5) & (df["Mental_Health_Index"] > 7)]
print(four_conditions)
display(four_conditions) # This will display the rows of the DataFrame where the daily usage hours are greater than 5 and the mental health score is less than 7
```

```python
five_conditions = df[(df["Daily_Usage_Hours"] > 6) & (df["Primary_Platform"] == "Instagram")]
print(five_conditions)
display(five_conditions) # This will display the rows of the DataFrame where the daily usage hours are greater than 6 and the primary platform is Instagram
```

```python
six_conditions = df[(df["Daily_Usage_Hours"] > 5) & (df["Sleep_Duration_Hours"] < 7) & (df["Mental_Health_Index"] > 5) & (df["Primary_Platform"] == "Youtube")]
print(six_conditions)
display(six_conditions) # This will display the rows of the DataFrame where the daily usage hours are greater than 6 and the sleep duration hours are less than 7
```

```python
group_by_age = df.groupby("Age")["Daily_Usage_Hours"].mean()

display(group_by_age) # This will display the average daily usage hours for each age group
```
